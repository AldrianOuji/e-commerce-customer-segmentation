```python
import pandas as pd

# Load the raw dataset using read_excel instead of read_csv
df = pd.read_excel('../data/Online_Retail.xlsx', engine='openpyxl')

# Look at the first 5 rows and column structures
print("✅ Excel data loaded successfully!")
print(df.info())
df.head()
```

```python
# 🧹 Check how many missing values (Nulls) are in each column
print("Missing values per column:")
print(df.isnull().sum())
```

```python
# 🧹 Dropping rows where CustomerID is blank
df_clean = df.dropna(subset=['CustomerID'])

# Double-check the new shape of your data
print(f"Original row count: {df.shape[0]}")
print(f"Cleaned row count: {df_clean.shape[0]}")
print(f"Removed rows: {df.shape[0] - df_clean.shape[0]}")
```

```python
# 🔍 Count how many transactions have negative quantities
cancelled_orders = df_clean[df_clean['Quantity'] <= 0]
print(f"Number of cancelled or returned rows: {cancelled_orders.shape[0]}")
```

```python
# 🚫 Keep only successful transactions where Quantity is greater than 0
df_clean = df_clean[df_clean['Quantity'] > 0]

# 💸 Also remove rare data entry glitches where UnitPrice is 0 or less
df_clean = df_clean[df_clean['UnitPrice'] > 0]

print(f"Final completely clean row count: {df_clean.shape[0]}")
```

```python
# 📊 Create a new column for the total monetary value of each transaction line item
df_clean['TotalSum'] = df_clean['Quantity'] * df_clean['UnitPrice']

# Review a small sample to confirm your math looks perfect
df_clean[['Quantity', 'UnitPrice', 'TotalSum']].head()
```
